```python
import os
import otter

from astropy.coordinates import SkyCoord
from astropy import units as u

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
db = otter.Otter()
```

```python
# initial function
# returns list of potential galaxy canidates

# initialize canidate list as objects classified as TDEs and within redshift 0.03
cands = db.query(classification="TDE", minz=0, maxz=0.03, class_confidence_threshold=1)

name_list = []
rs_list = []
cord_ra_list = []
cord_dec_list = []
hosts_info = []
confidence = []

i = 0
while i < len(cands):
    object = cands[i]
    
    name = object.default_name
    name_list.append(name)
    
    rs = float(object.get_redshift())
    rs_list.append(rs)
    
    cords = object.get_skycoord()
    ra = cords.ra.deg
    cord_ra_list.append(ra)
    dec = cords.dec.deg
    cord_dec_list.append(dec)

    conf = object.get_classification()
    confidence.append(conf[1])
    
    info = object.get_host()
    if info == []:
        hosts_info.append(False)
    else:
        hosts_info.append(True)

    i += 1



df = pd.DataFrame(
    {"Name": name_list, "redshift": rs_list, "confidence": confidence, "ra": cord_ra_list, "dec":cord_dec_list, "Host info in OTTER?": hosts_info}
)

df = df.set_index("Name")

print(df)
```
